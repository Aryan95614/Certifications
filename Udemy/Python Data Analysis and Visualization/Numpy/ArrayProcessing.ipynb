```python
import numpy as np
import matplotlib.pyplot as plt

# See the visual aspect
%matplotlib inline
```

```python
points = np.arange(start=-5, stop=5, step=.01)
```

```python
dx, dy = np.meshgrid(points, points)
```

```python
dx
```

```python
dy
```

```python
z = (np.sin(dx) + np.sin(dy))
z
```

```python
# To perform a visualization of a lot of data
# Use plt.imshow()

plt.imshow(z)
```

```python
plt.imshow(z)
plt.colorbar() # What do the colors mean

plt.title('Plot for sin x + sin y')
```

```python
# Using numpy where

A = np.array([1, 2, 3, 4])
B = np.array(A * 100)
```

```python
condition = np.array([True, True, False, False])
```

```python
# make an array where we choose A if true else choose B
# zip allows you to preload in multiple functions
answer = [(A_val if cond else B_val) for A_val, B_val, cond in zip(A, B, condition)]

answer
```

```python
# np.where() makes it so much easier
answer2 = np.where(condition, A, B)
```

```python
answer2
```
