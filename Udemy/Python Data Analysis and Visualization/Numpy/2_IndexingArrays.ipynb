```python
import numpy as np
```

```python
arr = np.arange(start=0, stop=11)
```

```python
print(arr)
```

```python
# Trivial Index
print(f" The 8th element is:\t {arr[8]}")
```

```python
# You can use list indexing
print(arr[1:5]) # [1 2 3 4]
print(arr[0:5]) # [0 1 2 3 4]
```

```python
# By using indexing, you can also assign values
arr[0:5] = 100
print(f"When changed through indexing, it becomes {arr}")

arr = np.arange(start=0, stop=11)
print(f" Changed it back into: {arr}")
```

```python
# Slicing arrays
slice_of_arr = arr[0:6] # also inherits the dtype

print(slice_of_arr)
```

```python
# Apply this to all elements
slice_of_arr[:] = 99
print(slice_of_arr)
```

```python
# Np array slices affect the thing itself
print(arr)
```

```python
# For no interference I have to do .copy()

arr_copy = arr.copy()
```

```python
# Making a 2d Array

arr_2d = np.array([[5, 10, 15], [20, 25, 30], [35, 40, 45]])
print(f"The 2d array becomes {arr_2d}")
```

```python
# Indexes
print(arr_2d[1])
```

```python
# Getting columns
print(arr_2d[1][0])
```

```python
# Creating a sub array within an array
print(arr_2d[:2, 1:]) # Everything but the last row and first column
```

```python
# Fancy Indexing
arr2d = np.zeros((10, 10))
print(arr2d)
```

```python
arr_length = arr2d.shape[1]
arr_length
```

```python
for i in range(arr_length):
    arr2d[i] = i
print(arr2d)
```

```python
# Getting the subarray again!!
print(arr2d[[2, 4, 6, 8]])
```
