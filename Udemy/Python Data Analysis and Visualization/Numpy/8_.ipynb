```python
import numpy as np  
```

```python
arr = np.arange(5)
```

```python
arr
```

```python
np.save('myarray', arr)
```

```python
np.load('myarray.npy')
```

```python
arr1 = np.load('myarray.npy')
```

```python
arr2 = arr
```

```python
arr2 
```

```python
np.savez('ziparray.npz', x=arr1, y=arr2)
```

```python
archive_array = np.load('ziparray.npz')
```

```python
# While these are the same you can make them
# different

print(archive_array['x'])
print(archive_array['y'])
```
