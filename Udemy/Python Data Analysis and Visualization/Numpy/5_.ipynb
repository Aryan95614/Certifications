```python
import numpy as np
```

```python
arr = np.arange(50).reshape((10, 5))

# print(f"Original Array:\t{np.arange(50)}")
print(f"Reshaped array:\n {arr}")
```

```python
print(arr.T)
```

```python
# Doing the dot product requires column of 1 and row of 2 
# to be the same
np.dot(arr.T, arr)
```

```python
# 3d matrix

arr3d = np.arange(50).reshape((5, 5, 2))
arr3d
```

```python
print(arr3d.transpose((1, 0, 2)))
```
