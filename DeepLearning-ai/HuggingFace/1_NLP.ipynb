```python
import transformers
from transformers import pipeline
```

```python
# the purpose of this is to create your own chatbot
chatbot = pipeline(task="conversational", model="facebook/blenderbot-400M-distill")
```

```python
user_message = "What are some fun activities that you can do in the winter?"
```

```python
from transformers import Conversation
```

```python
# to pass the message, you need to put it into a conversation object!
conversation = Conversation(user_message) # Creates an id and user

print(conversation)
```

```python
conversation=chatbot(conversation)
```
