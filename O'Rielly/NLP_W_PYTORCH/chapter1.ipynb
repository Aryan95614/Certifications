```python
import torch
import numpy as np
import pandas as pd
import matplotlib
```

```python
class Perceptron:
    """
    Parameters:
    ---
    eta: float
        The Learning Rate
    n_iter: int
        Pass over the training dataset
    random_state: int
        Random number generator seed fo rrandom weight initialization. 

    Attributes
    ---
    w_: 1d-array
       Weights after lifting
    b_: Scalar
        Bias unit after fitting

    errors:list
        How many misclassifications or epochs are there?
    """

    def __init__(self, eta=0.01, n_iter=50, random_state=1):
        self.eta = eta
        self.n_iter = n_iter
        self.random_state = random_state

    def fit(self, X, y):
        """
        This was created to fit the training data.

        Paramaters:
        ---
        X : (array-like), shape = [n_examples, n_features]
            These are training vectors, where n_examples is the amount
            of examples and n_features the amount of the required features
        y : array-like, shape = [n_examples]
            These are the target values

        Returns:
        ---
        It literally just returns itself
        """

        rgen = np.random.RandomState(self.random_state)
        self.w_ = rgen.normal(loc=0.0, scale=0.01, size=X.shape[1])
        self.b_ = np.float_(0.) # It literally is just a float balue
        self.errors_ = []

        for _ in range(self.n_iter):
            errors = 0
            for xi, target in zip(X, y):
                update = self.eta * (target - self.predict(xi))
                self.w_ += update * xi
                self.b_ += update 
                errors += int(update != 0.0) 
            self.errors_.append(errors)
        return self

    def net_input(self, X):
        """Calculates the net input"""
        return np.dot(X, self.w_) + self.b_

    def predict(self, X):
        """Return class label after unit step"""
        return np.where(self.net_input(X) >= 0.0, 1, 0)

```
